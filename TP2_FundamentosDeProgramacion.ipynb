##Ejercicio Nro 19: Escribir una función que calcule el área de un círculo y otra que calcule el volumen de un cilindro usando la primera función

###Área de un círculo
El área de un círculo es **$\pi = 3,14159265359$** (pi) multiplicado por el radio al cuadrado
<center >$AC = \pi*r^{2}$ </center>

para este ejercicio $\pi = 3,141592$

###Volumen de un cilindro
Para calcular el volumen de un cilindro hay que multiplicar el área de la base por la altura del cilindro. La base es circular, por lo que su área se calcula en $AC$
<center >$V = AC*h$ </center>

**Ejemplo**:

####Calculemos el área de un círculo

 Suponemos que el radio $r = 10 => AC = 3,141592 * 10^{2} = 3,141592 * 100 = 314,1592$

Entonces el Área de un círculo con radio 10 es $AC = 314,1592$
<br>
<br>
####Calculemos el volumen de un cilindro

 suponemos que la altura es de $h = 5$ y el área de su base es $AC = 314,1592$

 $V = 314,1592 * 5 = 1570,796$

 Entonces el Volumen de un cilindro es $V = 1570,796$

````python
#Resolver

PI = 3.141592

def area_circulo(radio):
  return PI * (radio ** 2)

def volumen_cilindro(radio, altura):
  return area_circulo(radio) * altura

def principal():
  radio = float(input("Ingrese un valor para el radio: "))
  altura = float(input("Ingrese un valor para la altura: "))
  resultado = volumen_cilindro(radio, altura)
  print(resultado)

principal()
````

##Ejercicio Nro 20: Escribir una función que calcule el máximo común divisor de dos números y otra que calcule el mínimo común múltiplo.

### Máximo común divisor
En matemáticas, se denomina máximo común divisor o MCD al mayor número que divide exactamente a dos o más números a la vez. Es decir resto igual a 0.

**Ejemplo**:

Supogamos que tengo el nnúmero 12 y 16:

*   Divisores de 12 son: 1, 2, 3, **4**, 6, 12 dividen a 12 sin dejar resto
*   Divisores de 16 son: 1, 2, **4**, 8, 16 dividen a 16 sin dejar resto

Pero el **MCD(12,16) = 4**  ya que es el número mas grande que divide ambos números sin dejar resto.

### Mínimo común múltiplo
En matemáticas, el mínimo común múltiplo de dos o más números naturales es el menor múltiplo común de todos ellos.

**Ejemplo**:

Supogamos que tengo el nnúmero 3 y 4:

*   Multiplos de 3 son: 3, 6, 9, **12**, 15, 18, 21, 24, 27.... son divisible por 3 sin dejar resto
*   Multiplos de 4 son: 4, 8, **12**, 16, 20, 24, 28, 32, 36.... son divisible por 4 sin dejar resto

Pero el **MCM(3,4) = 12**  ya que es el número mas chico que es divisible por ambos números sin dejar resto.

````python
#Resolver

def mcd(a, b):
  divisores_a = set() # lo hacemos de esta forma ya que si {}, Python va entender que estamos definiendo un diccionario vacio
  for i in range(1, a + 1): # 1 2 3 4 .... a - 1
    if a % i == 0:
      divisores_a.add(i)

  divisores_b = set()
  for i in range(1, b + 1):
    if b % i == 0:
      divisores_b.add(i)

  divisores_comunes = divisores_a & divisores_b # interseccion entre dos conjuntos
  lista = list(divisores_comunes)

  # deberia hacerlo con un metodo de ordenacion
  lista.sort()

  return lista[-1]

def mcm(a, b):
  if a == b:
    return a
  else:
    if a > b:
      mayor = a
      menor = b
    else:
      mayor = b
      menor = a

    i = 1
    while (mayor * i) % menor != 0:
      i = i + 1
    return mayor * i

def principal():
  num1 = 12
  num2 = 16
  print(f"El MCD de {num1} y {num2} es: {mcd(num1, num2)}")

  num3 = 3
  num4 = 4
  print(f"El MCM de {num3} y {num4} es: {mcm(num3, num4)}")

principal()
````

###**map**

Podemos usar la función integrada de Python **map()** para aplicar una función a cada elemento en un iterable (como una lista o un diccionario) y devolver un nuevo iterador para recuperar los resultados. **map()** devuelve un objeto de asignación (un iterador) que podemos usar en otras partes de nuestro programa.

**Ejemplo**

````python
#Uso de map
palabras = ['hola','mundo','esto', 'es', 'Python']
largo = map(len,palabras)
c = 0
for i in largo:
  print(f"la longitud de {palabras[c]} es: {i}")
  c +=1
````

##**zip**

La función incorporada **zip()** toma como argumento dos o más objetos iterables (idealmente cada uno de ellos con la misma cantidad de elementos) y retorna un nuevo iterable cuyos elementos son tuplas que contienen un elemento de cada uno de los iteradores originales.

**Ejemplo**

````python
palabras = ['hola','mundo','esto', 'es', 'Python']
largo = map(len,palabras)
diccionario = dict(zip(palabras,largo))
print(diccionario)
````

````python
#Resolver

def contar_palabra(frase):
  palabras = frase.split(" ")
  largos = map(len, palabras) # [4, 5, 4, 2, 6]
  # [('hola', 4), ('mundo', 5), ...]
  diccionario = dict(zip(palabras, largos))
  return diccionario

def principal():
  frase = "Esto es Python"
  print(contar_palabra(frase))

principal()
````

## Ejercicio Nro 25: Escriba una función calcular(), que puede recibir cualquier cantidad de números y devolver una tupla.

El primer valor de la tupla es el valor promedio de todos los parámetros y el segundo valor son todos los números mayores que el valor promedio

**Entrada**: 5,8,10,2,3,7

**Salida**: (5.83,[8,10,7])

**Entrada**: 5,5

**Salida**: (5,[])

````python
#Resolver

def calcular(*args):
  acumulador = 0
  for elemento in args:
    acumulador = acumulador + elemento
  promedio =  acumulador / len(args)
  lista = []
  for elemento in args:
    if elemento > promedio:
      lista.append(elemento)
  return promedio, lista
````

Los inmuebles de la lista que se devuelva deben incorporar un nuevo par a cada diccionario con el precio del inmueble, donde el precio de un inmueble se calcula con las siguiente fórmula en función de la zona:

- Zona A: precio = (metros x 1000 + habitaciones x 5000 + garaje x 15000) x (1 - antiguedad / 100)
- Zona B: precio = (metros x 1000 + habitaciones x 5000 + garaje x 15000) x (1 - antiguedad / 100) x 1.5
