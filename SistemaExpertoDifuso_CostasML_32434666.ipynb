```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

# Variables de entrada
color = ctrl.Antecedent(np.arange(0, 11, 1), 'color')       # escala 0 (claro) a 10 (oscuro)
brillo = ctrl.Antecedent(np.arange(0, 11, 1), 'brillo')     # escala 0 (no metálico) a 10 (metálico)
dureza = ctrl.Antecedent(np.arange(0, 11, 1), 'dureza')     # escala 0 (muy blando) a 10 (duro)

# Variable de salida
mineral = ctrl.Consequent(np.arange(0, 11, 1), 'mineral')

# Funciones de pertenencia
color['claro'] = fuzz.trimf(color.universe, [0, 0, 5])
color['medio'] = fuzz.trimf(color.universe, [2, 5, 8])
color['oscuro'] = fuzz.trimf(color.universe, [5, 10, 10])

brillo['no_metalico'] = fuzz.trimf(brillo.universe, [0, 0, 5])
brillo['medio'] = fuzz.trimf(brillo.universe, [2, 5, 8])
brillo['metalico'] = fuzz.trimf(brillo.universe, [5, 10, 10])

dureza['muy_blando'] = fuzz.trimf(dureza.universe, [0, 0, 3])
dureza['blando'] = fuzz.trimf(dureza.universe, [2, 4, 6])
dureza['medio'] = fuzz.trimf(dureza.universe, [5, 6, 7])
dureza['duro'] = fuzz.trimf(dureza.universe, [7, 10, 10])

mineral['talco'] = fuzz.trimf(mineral.universe, [0, 0, 2])
mineral['yeso'] = fuzz.trimf(mineral.universe, [2, 3, 4])
mineral['baritina'] = fuzz.trimf(mineral.universe, [4, 5, 6])
mineral['cuarzo'] = fuzz.trimf(mineral.universe, [6, 8, 10])

# REGLAS DIFUSAS
regla1 = ctrl.Rule(color['claro'] & brillo['no_metalico'] & dureza['muy_blando'], mineral['talco'])
regla2 = ctrl.Rule(color['claro'] & brillo['no_metalico'] & dureza['blando'], mineral['yeso'])
regla3 = ctrl.Rule(color['claro'] & brillo['no_metalico'] & dureza['medio'], mineral['baritina'])
regla4 = ctrl.Rule(color['oscuro'] & brillo['metalico'] & dureza['duro'], mineral['cuarzo'])

# MOTOR DE INFERENCIA
sistema_mineral = ctrl.ControlSystem([regla1, regla2, regla3, regla4])
simulador = ctrl.ControlSystemSimulation(sistema_mineral)

# INTERFAZ
def iniciar_identificacion_difusa():
    print("== Sistema Experto Difuso para Reconocimiento de Minerales ==")
    print("Ingrese valores del 0 al 10 para cada característica.\n")

    try:
        simulador.input['color'] = float(input("Color (0=claro, 10=oscuro): "))
        simulador.input['brillo'] = float(input("Brillo (0=no metálico, 10=metálico): "))
        simulador.input['dureza'] = float(input("Dureza (0=muy blando, 10=duro): "))

        simulador.compute()

        if 'mineral' in simulador.output:
            resultado = simulador.output['mineral']
            print(f"\nValor difuso de salida: {resultado:.2f}")

            if resultado < 2:
                print("→ Mineral probable: Talco")
            elif resultado < 4:
                print("→ Mineral probable: Yeso")
            elif resultado < 6:
                print("→ Mineral probable: Baritina")
            else:
                print("→ Mineral probable: Cuarzo")
        else:
            print("\nNo se pudo determinar el mineral con los valores ingresados. Intente con otros valores.")

    except ValueError:
        print("\nEntrada inválida. Por favor, ingrese valores numéricos.")
    except Exception as e:
        print(f"\nOcurrió un error: {e}")

iniciar_identificacion_difusa()
```
