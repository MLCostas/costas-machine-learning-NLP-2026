```python
# 1. BASE DE CONOCIMIENTO (Knowledge Base)
# Contiene el conocimiento del dominio en forma de reglas [1-3].
# Cada regla es una unidad de conocimiento modular e independiente [2].
base_conocimiento = {
        'Regla 61': {
        'condiciones': {'color': 'amarillo', 'brillo': 'metalico', 'dureza': 'medio'},
        'conclusion': 'Oro'
    },
    'Regla 2': {
        'condiciones': {'color': 'gris plateado', 'brillo': 'metalico', 'dureza': 'duro'},
        'conclusion': 'Plata'
    },
    'Regla 3': {
        'condiciones': {'color': 'rojo', 'brillo': 'metalico', 'dureza': 'duro'},
        'conclusion': 'Cobre'
    },
    'Regla 4': {
        'condiciones': {'color': 'negro', 'brillo': 'no metalico', 'dureza': 'medio'},
        'conclusion': 'Uranio (forma mineral: uraninita)'
    },
    'Regla 5': {
        'condiciones': {'color': 'gris', 'brillo': 'metalico', 'dureza': 'duro'},
        'conclusion': 'Hierro (forma mineral: hematita)'
    },
    'Regla 6': {
        'condiciones': {'color': 'gris plateado', 'brillo': 'metalico', 'dureza': 'medio'},
        'conclusion': 'Molibdeno (forma mineral: molibdenita)'
    },
    'Regla 7': {
        'condiciones': {'color': 'blanco', 'brillo': 'no metalico', 'dureza': 'blando'},
        'conclusion': 'Boratos (forma mineral: borax)'
    },
    'Regla 8': {
        'condiciones': {'color': 'blanco', 'brillo': 'no metalico', 'sabor': 'salado'},
        'conclusion': 'Sal (cloruro de sodio)'
    },
    'Regla 9': {
        'condiciones': {'color': 'blanco', 'brillo': 'no metalico', 'dureza': 'blando'},
        'conclusion': 'Yeso'
    },
    'Regla 10': {
        'condiciones': {'color': 'verde', 'brillo': 'no metalico', 'dureza': 'medio'},
        'conclusion': 'Fluorita'
    },
    'Regla 11': {
        'condiciones': {'color': 'blanco', 'brillo': 'no metalico', 'dureza': 'medio'},
        'conclusion': 'Baritina'
    },
    'Regla 12': {
        'condiciones': {'color': 'blanco', 'brillo': 'mate', 'dureza': 'blando'},
        'conclusion': 'Caolín'
    },
    'Regla 13': {
        'condiciones': {'color': 'rosado', 'brillo': 'no metalico', 'dureza': 'medio'},
        'conclusion': 'Rodocrosita'
    },
    'Regla 14': {
        'condiciones': {'color': 'violeta', 'brillo': 'no metalico', 'dureza': 'duro'},
        'conclusion': 'Amatista'
    },
    'Regla 15': {
        'condiciones': {'color': 'verde', 'brillo': 'no metalico', 'dureza': 'duro'},
        'conclusion': 'Turmalina'
    },
    'Regla 16': {
        'condiciones': {'color': 'incoloro', 'brillo': 'vítreo', 'dureza': 'duro'},
        'conclusion': 'Cuarzo cristal'
    },
    'Regla 17': {
        'condiciones': {'color': 'blanco', 'brillo': 'vítreo', 'dureza': 'medio'},
        'conclusion': 'Feldespato'
    },
    'Regla 18': {
        'condiciones': {'color': 'blanco', 'brillo': 'no metalico', 'dureza': 'muy blando', 'reaccion_hcl': 'no'},
        'conclusion': 'Talco'
    }
}

# 2. MOTOR DE INFERENCIA (Inference Engine)
# Componente que "razona" aplicando las reglas a los hechos [3, 4, 6].
# Este motor utiliza una estrategia de encadenamiento hacia adelante (Forward Chaining),
# ya que parte de los datos conocidos para llegar a una conclusión [6-8].
def motor_de_inferencia(hechos_usuario, base_conocimiento):
    for nombre_regla, contenido_regla in base_conocimiento.items():
        condiciones = contenido_regla['condiciones']
        conclusion = contenido_regla['conclusion']

        regla_cumplida = True

        for propiedad, valor_esperado in condiciones.items():
            if propiedad not in hechos_usuario:
                regla_cumplida = False
                break

            valor_usuario = hechos_usuario[propiedad]

            if isinstance(valor_esperado, list):
                if valor_usuario not in valor_esperado:
                    regla_cumplida = False
                    break
            else:
                if valor_usuario != valor_esperado:
                    regla_cumplida = False
                    break

        if regla_cumplida:
            return conclusion

    return "No se pudo identificar el mineral con la información proporcionada."

# 3. INTERFAZ DE USUARIO (User Interface)
# Permite la interacción entre el usuario y el sistema experto [3, 5].
def iniciar_identificacion():
    """
    Función principal que maneja la interacción con el usuario para
    recopilar hechos y presentar una conclusión.
    """
    print("======================================================")
    print("== Asistente para Identificación Preliminar de Minerales ==")
    print("======================================================")
    print("Por favor, responda a las siguientes preguntas sobre el mineral, siguiendo el protocolo de reconocimiento de minerales.")
    print("Escriba su respuesta en minúsculas y presione Enter.\n")

    hechos_usuario = {}

    # Proceso de adquisición de conocimiento a través de preguntas al usuario [9].

    color = input("-> ¿Cuál es el color del mineral? (ej: blanco, verde, amarillo laton, incoloro): ")
    hechos_usuario['color'] = color.strip()

    brillo = input("-> ¿El brillo es metálico o no metálico? (escriba 'metalico' o 'no metalico'): ")
    hechos_usuario['brillo'] = brillo.strip()

    print("\nPara determinar la dureza, intente rayar el mineral:")
    dureza = input("-> ¿Se raya con la uña ('muy blando'), con una navaja de acero ('blando' o 'medio'), o raya el acero ('duro')?: ")
    hechos_usuario['dureza'] = dureza.strip()

    reaccion = input("-> ¿El mineral efervesce (burbujea) con ácido? (escriba 'efervesce' o 'no'): ")
    if reaccion.strip() == 'efervesce':
        hechos_usuario['reaccion_hcl'] = 'efervesce'
    else:
        hechos_usuario['reaccion_hcl'] = 'no'


    # Llamada al motor de inferencia con los datos recopilados [10].
    print("\n------------------------------------------------------")
    print("Analizando la información...")
    diagnostico = motor_de_inferencia(hechos_usuario, base_conocimiento)

    # Presentación de resultados de forma comprensible [5].
    print("\n** Diagnóstico del Sistema Experto **")
    print(f"Hechos proporcionados: {hechos_usuario}")
    print(f"Conclusión: El mineral podría ser **{diagnostico}**.")

    # Muchos sistemas expertos pueden explicar el razonamiento detrás de sus conclusiones,
    # en este caso, mostramos los hechos que llevaron al resultado [11].
    if "No se pudo identificar" not in diagnostico:
        print("\nRecomendación: Para una identificación definitiva, se recomienda validar estas características en una guía de minerales y, si es posible, consultar a un geólogo.")
    else:
        print("\nRecomendación: Intente de nuevo con características más definidas o consulte una guía especializada.")

# --- Ejecutar el sistema ---
if __name__ == "__main__":
    iniciar_identificacion()
```
